#### Problem Statement:  
  
You are given a 0-indexed integer array nums of even length consisting of an equal number of positive and negative integers.  
  
You should return the array of nums such that the the array follows the given conditions:  
  
Every consecutive pair of integers have opposite signs.  
For all integers with the same sign, the order in which they were present in nums is preserved.  
The rearranged array begins with a positive integer.  
Return the modified array after rearranging the elements to satisfy the aforementioned conditions.  
  
   
  
Example 1:  
 
Input: nums = [3,1,-2,-5,2,-4]  
Output: [3,-2,1,-5,2,-4] 
 
Explanation:  
The positive integers in nums are [3,1,2]. The negative integers are [-2,-5,-4].  
The only possible way to rearrange them such that they satisfy all conditions is [3,-2,1,-5,2,-4].  
Other ways such as [1,-2,2,-5,3,-4], [3,1,2,-2,-5,-4], [-2,3,-5,1,-4,2] are incorrect because they do not satisfy one or more conditions.    
Example 2:  
  
Input: nums = [-1,1]  
Output: [1,-1]  
Explanation:  
1 is the only positive integer and -1 the only negative integer in nums.  
So nums is rearranged to [1,-1].  
  
   
  
Constraints:  
  
2 <= nums.length <= 2 * 105  
nums.length is even  
1 <= |nums[i]| <= 105  
nums consists of equal number of positive and negative integers.  
   
  
It is not required to do the modifications in-place.

#### Brute Solution:
The solution separates positive and negative numbers into two arrays and then merges them in alternating order into the result array. 🚀

```python
%%script node

var rearrangeArray = function(nums) {
    let n = nums.length;

    // Arrays to store positive and negative numbers separately
    let posEle = [];
    let negEle = [];
    
    // Result array of the same length as input
    let result = new Array(n);

    // Separate positive and negative numbers into different arrays
    for (let i = 0; i < n; i++) {
        if (nums[i] > 0) {
            posEle.push(nums[i]);
        } else {
            negEle.push(nums[i]);
        }
    }

    // Interleave positive and negative numbers in the result array
    for (let i = 0; i < Math.floor(n / 2); i++) {
        result[2 * i] = posEle[i];       // Assign positive number at even index
        result[2 * i + 1] = negEle[i];   // Assign negative number at odd index
    }

    return result;
};

// Example usage:
console.log(rearrangeArray([3,1,-2,-5,2,-4]));  
// Expected Output: [3, -2, 1, -5, 2, -4]

/* 
📌 Time Complexity: O(n) → We iterate through the array twice (once to separate, once to merge).  
📌 Space Complexity: O(n) → We use extra space for two arrays (posEle & negEle) and the result array.  
🚀 A straightforward and easy-to-understand approach!  
*/
```

#### Optimal Solution
The solution places positive and negative numbers alternately in a new array using two pointers for even and odd indices. 🚀

```python
%%script node

// Function to rearrange an array such that positive and negative numbers are placed alternately
var rearrangeArray = function(nums) {
    let n = nums.length;
    
    // Create a new array of the same size to store the rearranged elements
    let result = Array(n);

    // Pointers to track the next available positions for positive and negative numbers
    let positiveInd = 0, negativeInd = 1;
    
    // Iterate through the original array
    for (let i = 0; i < n; i++) {
        if (nums[i] > 0) {
            // Place positive number at even index
            result[positiveInd] = nums[i];
            positiveInd += 2;
        } else {
            // Place negative number at odd index
            result[negativeInd] = nums[i];
            negativeInd += 2;
        }
    }

    // Return the rearranged array
    return result;
};

// Example usage:
console.log(rearrangeArray([3,1,-2,-5,2,-4]));  
// Expected Output: [3, -2, 1, -5, 2, -4]

/* 
📌 Time Complexity: O(n) → We iterate through the array once.  
📌 Space Complexity: O(n) → We use just an additional array of the same size.  
🚀 Efficient and simple approach!
*/
```
